# Final Product

By the end of this workshop, we will have a basic **investment portfolio manager**.



````python
def manage_portfolio(market_opportunities, budget):
    '''
    Allocates capital to market opportunities based on a given budget.
    Iterates through available market opportunities and makes purchases
    if there is available budget.
    '''
    purchased = []

    print(f"Starting Budget: ${budget}")

    for ticker in market_opportunities:
        price = market_opportunities[ticker]

        # Decision Logic: If we can afford it, buy it
        if price <= budget:
            purchased.append(ticker)
            budget -= price
            print(f"Decision: BUY. Bought 1 share of {ticker} for ${price}")
        else:
            print(f"Decision: HOLD. Skipped {ticker}: Price ${price} exceeds remaining budget.")

    print(f"Remaining Cash: ${budget}")
    print(f"Portfolio Assets: {purchased}")

# Market Data: {Ticker: Share Price}
opportunities = {'NVDA': 450.00, 'GOOGL': 130.00, 'AMZN': 145.00, 'META': 300.00}

user_budget = float(1000)
manage_portfolio(opportunities, user_budget)
````

# 1.2 f-strings

f-strings (f-notation) allow you to embed variables and expressions directly into strings by prefixing the text with an `f` and wrapping the variable in curly braces `{}`.

````python
# Our earlier function used an f-string to print day of the week
print(f'Happy {day_of_the_week}!')
````

F-string notation also supports the formatting of numbers.

The structure for number formatting is `f"{value:{format_specifier}}"`, where the format specifier is introduced by a colon (`:`)

A common format specifier is choosing significant number of digits.

```
pi = 3.1415926535
print(f"Standard: {pi:.2f}")  # Rounds to 2 decimal places: 3.14
print(f"Precise: {pi:.4f}")   # Rounds to 4 decimal places: 3.1416
```

````python
# Example: math inside an f-string!
quantity = 90
item = 'eggs'
price = 1.99

print(f"The value of {quantity} {item} is ${quantity * price:0.2f}")
````

## 1.4 Global vs. Local Variables

Variable scope determines where a variable can be seen or used within your code. Depending on where it is defined, a variable is either **global** or **local**.

Variables created outside of any function are **global**.

Variables created inside a function are **local**. Local variables are not accessible from outside of their defined functions.

## 1.9 Let's put it together!

````python
# Combining a for loop, lists, dict, and an if statement
def buy_snacks(snack_options, budget):
  # An empty list
  purchased_snacks = []

  # Printing our budget
  print(f"--- Starting Budget: ${budget}--- ")

  # Printing our snack options
  print(f"--- Snack Options: {snack_options}")

  for snack in snack_options:
    price = snack_options[snack]

    # if we have the budget, we will buy the snack
    if price <= snack_budget:
      # buy
      purchased_snacks.append(snack)
      snack_budget -= snack
      print(f"We bought {snack} at price ${price}.")
    else:
      # hold
      print(f"We did not buy {snack} at price ${price}, it exceeded our budget of {snack_budget}")

  print(f"Remaining Cash: ${budget:.2f}")
  print(f"Purchased Snacks: {purchased_snacks}")
````
